## 1 · Versión de Python y librerías

```python
import sys, platform
print('Python', sys.version.split()[0], '·', platform.system())

# El nombre con el que se importa no siempre es el nombre con el que se instala.
librerias = {'pandas': 'pandas', 'numpy': 'numpy', 'sklearn': 'scikit-learn',
             'matplotlib': 'matplotlib', 'requests': 'requests'}

for importar, instalar in librerias.items():
    try:
        mod = __import__(importar)
        print(f'  OK    {importar:<12} {getattr(mod, "__version__", "?")}')
    except ImportError:
        print(f'  FALTA {importar:<12} ->  pip install {instalar}')
```

## 2 · ¿Estamos en Colab, en Codespaces o en local?

Importa saberlo, porque el sitio donde viven los ficheros cambia.

```python
import os

try:
    shell = str(get_ipython())
except NameError:   # fuera de un kernel de notebook
    shell = ''

if 'google.colab' in shell:
    entorno = 'Google Colab'
elif os.environ.get('CODESPACES') == 'true':
    entorno = 'GitHub Codespaces'
else:
    entorno = 'Local'

print('Entorno:', entorno)
print('Directorio de trabajo:', os.getcwd())
```

## 3 · La estructura de carpetas del proyecto

Las cuatro capas de datos son las de la diapositiva del modelo de tres capas,
más una carpeta `sample` con la muestra que sí se versiona.

```python
from pathlib import Path

raiz = Path.cwd()
if raiz.name == 'notebooks':
    raiz = raiz.parent

esperadas = ['data/raw', 'data/sample', 'data/clean', 'data/analytic',
             'notebooks', 'docs']
for c in esperadas:
    existe = (raiz / c).is_dir()
    print(('  OK   ' if existe else '  FALTA ') + c)
```

## 4 · Las claves no están en el repositorio

Un token es una contraseña. Vive en `.env`, que está en `.gitignore`.
Esta celda comprueba que **no** habéis subido claves por error.

```python
from pathlib import Path

raiz = Path.cwd()
if raiz.name == 'notebooks':
    raiz = raiz.parent

gitignore = (raiz / '.gitignore')
protegido = gitignore.exists() and '.env' in gitignore.read_text(encoding='utf-8')
print(('  OK   ' if protegido else '  AVISO ') + '.env está en .gitignore')

if (raiz / '.env').exists():
    print('  OK   existe un .env local (no se sube)')
else:
    print('  NOTA  todavía no hay .env: copiad .env.example cuando tengáis claves')
```

## 5 · Leer y escribir datos

Última comprobación: que se puede crear un DataFrame y guardarlo donde toca.

```python
import pandas as pd
from pathlib import Path

raiz = Path.cwd()
if raiz.name == 'notebooks':
    raiz = raiz.parent

df = pd.DataFrame({
    'texto': ['El envío llegó tarde pero el producto es correcto.',
              'Atención al cliente impecable, lo recomiendo.'],
    'fecha': ['2026-09-10', '2026-09-10'],
    'url':   ['https://ejemplo.test/1', 'https://ejemplo.test/2'],
})

destino = raiz / 'data' / 'sample' / 'prueba.csv'
df.to_csv(destino, index=False, encoding='utf-8')
print('  OK   escrito en', destino.relative_to(raiz))
print()
print(pd.read_csv(destino))
```
